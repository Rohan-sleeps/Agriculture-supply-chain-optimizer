## Inspecting Mandi Master before cleaning it

```python

df_master_clean = df_master.copy(deep=True)


display(df_master_clean)

# Inspect the exact spellings and formats.
# repr() makes spaces and empty strings easier to see.
for column in ["mandi_id", "district", "state", "mandi_type"]:
    print(f"\nUnique values in {column}:")
    print([repr(value) for value in df_master_clean[column].unique()])

# Show every occurrence of an exact duplicate.
print("\nExact duplicate groups:")
display(
    df_master_clean[
        df_master_clean.duplicated(keep=False)
    ].sort_values("mandi_id")
)
```

## Cleaning the copy of our data

```python
# Use only records with both district and state present.
known_locations = (
    df_master_clean[["district", "state"]]
    .dropna()
    .drop_duplicates()
)

# Check whether any district is associated with multiple states.
states_per_district = (
    known_locations.groupby("district")["state"].nunique()
)

# Only use districts with exactly one observed state.
safe_districts = states_per_district[
    states_per_district == 1
].index

district_to_state = (
    known_locations[
        known_locations["district"].isin(safe_districts)
    ]
    .set_index("district")["state"]
)

# Identify missing states that can be recovered.
suggested_state = df_master_clean["district"].map(
    district_to_state
)

fill_mask = (
    df_master_clean["state"].isna()
    & suggested_state.notna()
)

# Keep a record of each correction.
state_fill_log = df_master_clean.loc[
    fill_mask, ["mandi_id", "district", "state"]
].copy()

state_fill_log = state_fill_log.rename(
    columns={"state": "original_state"}
)

state_fill_log["filled_state"] = suggested_state.loc[fill_mask]
state_fill_log["reason"] = (
    "Unique district-to-state relationship in mandi master"
)

# Fill only the recoverable missing states.
df_master_clean.loc[fill_mask, "state"] = (
    suggested_state.loc[fill_mask]
)

print("Missing states filled:", int(fill_mask.sum()))
print("Missing states remaining:", df_master_clean["state"].isna().sum())

display(state_fill_log)
```

## check whether the prices dataset can help recover the four missing districts. Prices contain both mandi_id and district, but we must check for conflicting mappings before using them.

```python
# Identify mandis whose district is still missing.
missing_districts = df_master_clean.loc[
    df_master_clean["district"].isna(),
    ["mandi_id", "mandi_name", "state"]
]

display(missing_districts)

# Prepare a temporary lookup from prices.
price_locations = df_prices[["mandi_id", "district"]].copy()

# Normalize ID formats such as MANDI-001, mandi_001 and M001.
id_text = (
    price_locations["mandi_id"]
    .astype("string")
    .str.strip()
    .str.upper()
    .str.replace(r"[\s_-]+", "", regex=True)
)

# Accept only a complete recognized ID pattern.
id_number = id_text.str.extract(
    r"^(?:MANDI|M)?(\d{1,3})$",
    expand=False
)

price_locations["mandi_id"] = (
    "MANDI" + id_number.str.zfill(3)
)

price_locations["district"] = (
    price_locations["district"]
    .astype("string")
    .str.strip()
    .str.title()
    .replace("", pd.NA)
)

# Inspect district evidence for only the affected mandis.
district_evidence = (
    price_locations.loc[
        price_locations["mandi_id"].isin(
            missing_districts["mandi_id"]
        )
    ]
    .dropna(subset=["mandi_id", "district"])
    .groupby(["mandi_id", "district"])
    .size()
    .reset_index(name="price_record_count")
    .sort_values(
        ["mandi_id", "price_record_count"],
        ascending=[True, False]
    )
)

display(district_evidence)
```

## This shows we cannot reliably recover these districts as they have confliciting evidences we are keeping this four districts as missing and flag them as unresolved

```python
df_master_clean["district_status"] = "Present in master"

missing_mask = df_master_clean["district"].isna()

df_master_clean.loc[
    missing_mask, "district_status"
] = "Unresolved: conflicting price districts"

district_issue_log = df_master_clean.loc[
    missing_mask,
    ["mandi_id", "mandi_name", "state", "district_status"]
].copy()

display(district_issue_log)
```

## Checking the remaining numeric field

```python
area_raw = df_master_clean["total_area_acres"]

area_numeric = pd.to_numeric(area_raw, errors="coerce")

df_master_clean["area_status"] = "Valid"

df_master_clean.loc[
    area_raw.isna(), "area_status"
] = "Missing"

df_master_clean.loc[
    area_raw.notna() & area_numeric.isna(),
    "area_status"
] = "Invalid numeric value"

df_master_clean.loc[
    area_numeric.le(0), "area_status"
] = "Nonpositive area: review"

df_master_clean["total_area_acres"] = area_numeric

display(
    df_master_clean["area_status"]
    .value_counts()
    .to_frame("mandi_count")
)
```
